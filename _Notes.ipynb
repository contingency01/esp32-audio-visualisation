## Pin connections
#### I²S mic

- L/R → GND (left channel selected)

- WS → GPIO25

- SCK → GPIO14

- SD → GPIO34

- VDD → 3V3

- GND → GND

#### TFT LCD (SPI, 1.8" 128×160)

- GND → GND

- VCC → 3V3

- SCL → GPIO27 (SPI clock, SCK)

- SDA → GPIO33 (SPI MOSI)

- RES → GPIO12

- DC → GPIO32

- CS → GPIO26

- BL → (planned: 3V3, or a GPIO later)

#### Some <code>mpremote</code> tricks:

List of files on the board: <code> mpremote connect /dev/tty.usbserial-10 ls </code> 

Open REPL after copying: <code> mpremote connect /dev/tty.usbserial-10 cp main.py :main.py repl</code> 

Soft reset from REPL: <code> import machine; machine.reset() </code> 




## Project structure
We write/edit:

- audio.py (I2S mic interface)

- display.py (TFT drawing)

- processing.py

- config.py (pins)

- main.py (main loop)


We either:

- <code>mpremote ... reset</code> to auto-run main.py, or

- <code>mpremote ... run main.py</code> during development.
